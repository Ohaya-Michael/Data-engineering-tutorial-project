## 1. Write checks as a small test notebook:

```python
checks = {
  "no_null_year":     "SELECT COUNT(*) FROM roadsafe.silver.accidents WHERE year IS NULL",
  "valid_severity":   "SELECT COUNT(*) FROM roadsafe.silver.accidents WHERE severity IS NULL",
  "hour_in_range":    "SELECT COUNT(*) FROM roadsafe.silver.accidents WHERE hour NOT BETWEEN 0 AND 23",
  "month_in_range":   "SELECT COUNT(*) FROM roadsafe.silver.accidents WHERE month NOT BETWEEN 1 AND 12",
}

failed = []
for name, sql in checks.items():
    bad = spark.sql(sql).collect()[0][0]
    print(f"{name}: {bad} bad rows")
    if bad > 0:
        failed.append(name)

assert not failed, f"Quality checks failed: {failed}"
```
